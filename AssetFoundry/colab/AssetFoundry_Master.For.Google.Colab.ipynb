````python
#@title Generate hundreds of focused Colab notebooks
!rm -rf AssetFoundry/Notebooks
!python AssetFoundry/tools/generate_asset_notebooks.py --output AssetFoundry/Notebooks --repository "$REPOSITORY" --branch "$BRANCH"
notebooks = sorted(Path('AssetFoundry/Notebooks').rglob('*.ipynb'))
print('Generated notebooks:', len(notebooks))
assert len(notebooks) == catalog['asset_count']
````

````python
#@title Browse asset IDs and approved Godot destinations
CATEGORY = 'character' #@param ['architecture','environment','foliage','character','vehicle','defense','props','ui','vfx']
matches = [entry for entry in catalog['assets'] if entry['category'] == CATEGORY]
for entry in matches[:100]:
    print(entry['asset_id'], '=>', entry['godot_destination'], '|', entry['backend'])
print('Category count:', len(matches))
````

## Mesh and topology contract

Focused backends share a continuous mesh-state contract `z_i=[p_i,n_i,e_i]`. Candidate connectivity is decoded from spacetime distance, then checked for manifoldness, winding, components, UV integrity, scale, LOD error, collision suitability, rig deformation, animation completeness, and Godot import behavior. The repository implements compatible contracts and deterministic refinement stages; it does not claim to include unreleased research weights.


````python
#@title Package the generated notebook suite for manual use
import shutil
archive = shutil.make_archive('/content/ZomNom_AssetFoundry_Notebooks', 'zip', 'AssetFoundry/Notebooks')
print(archive)
from google.colab import files
files.download(archive)
````

## After running focused notebooks

Extract each downloaded bundle beneath `incoming/<asset_id>/`, clone the repository locally, and run:

```bash
python AssetFoundry/tools/mapping_surfer.py --incoming incoming --output AssetFoundry/output/godot_patch --zip AssetFoundry/output/zom_nom_asset_patch.zip
```

The mapping surfacer ignores bundle-provided destinations and resolves placement from the repository catalog.
